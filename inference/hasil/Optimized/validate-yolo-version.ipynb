```python

# Fungsi untuk mengunduh file dari Google Drive
def download_file_from_google_drive(url, output_path):
    # Mengambil ID file dari URL
    file_id = url.split('/')[-2]
    download_url = f"https://drive.google.com/uc?id={file_id}"

    # Mengunduh file menggunakan gdown
    gdown.download(download_url, output_path, quiet=False)

# Fungsi untuk mengekstrak file ZIP
def unzip_file(zip_path, extract_to):
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
        print(f"File diekstrak ke {extract_to}")

# Fungsi untuk memuat model YOLO
def load_yolo_model(model_path):
    # Memuat model YOLO dari file .pt
    model = YOLO(model_path)
    print("Model berhasil dimuat.")
    return model

# Fungsi untuk validasi model YOLO
def validate_yolo_model(model, data_yaml_path):
    results = model.val(data=data_yaml_path, task='test', save=True)
    return results
```

```python
import os

model_directory = '/content/model/All model'
model_files = []
if os.path.exists(model_directory):
    print(f"Contents of '{model_directory}':")
    for item in os.listdir(model_directory):
        model_files.append(item)
        print(item)
else:
    print(f"The directory '{model_directory}' does not exist.")
```

```python
import re

def extract_version(filename):
    match = re.search(r'yolov(\d+)', filename)
    return int(match.group(1)) if match else 999

model_files = sorted(model_files, key=lambda x: (extract_version(x), x))
print("\n".join(model_files))
```

```python
def evaluate_multiple_models(models_dir, data_yaml_path):

    for model_file in model_files:
        model_path = models_dir + "/" + model_file  # Concatenate path
        print(model_path)
        if os.path.exists(model_path):
            print(f"{'-'*10} {model_file} {'-'*10}")
            model = load_yolo_model(model_path)
            validate_yolo_model(model, data_yaml_path)
        else:
            print(f"Model {model_file} tidak ditemukan di direktori {models_dir}")
```

```python
models_dir = '/content/model/All model'

# Path ke file YAML dataset
data_yaml_path = '/content/Armed-Person-Recognition-4/data.yaml'

# Evaluasi model-model YOLO
evaluate_multiple_models(models_dir, data_yaml_path)
```
